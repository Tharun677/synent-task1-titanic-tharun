```python
# Import pandas and numpy libraries for data handling
import pandas as pd
import numpy as np
```

```python
# Load the Titanic dataset into a DataFrame
df = pd.read_csv('Titanic-Dataset.csv')
```

```python
# Display the first 5 rows of the dataset
df.head()
```

```python
# Show the number of rows and columns
df.shape
```

```python
# Display all column names
df.columns
```

```python
# Show dataset information including data types and null values
df.info()
```

```python
# Count missing values in each column
df.isnull().sum()
```

```python
# Replace missing ages with the median age
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
# Replace missing embarkation values with the most common value
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
# Calculate percentage of missing values in Cabin column
df['Cabin'].isnull().mean()*100
```

```python
# Drop Cabin column because it contains too many missing values
df.drop('Cabin', axis=1, inplace=True)
```

```python
# Confirm that all missing values have been handled
df.isnull().sum()
```

```python
# Count duplicate rows in the dataset
df.duplicated().sum()
```

```python
# Remove duplicate rows from the dataset
df.drop_duplicates(inplace=True)
```

```python
# Confirm that no duplicate rows remain
df.duplicated().sum()
```

```python
# Check dataset size after removing duplicates
df.shape
```

```python
# Rename columns to improve readability
df.rename(columns={
    'PassengerId': 'Passenger_ID',
    'Pclass': 'Passenger_Class',
    'Sex': 'Gender'
}, inplace=True)
```

```python
# Display updated column names
df.columns
```

```python
# Display current data types of all columns
df.dtypes
```

```python
# Convert categorical columns to category data type
df['Survived'] = df['Survived'].astype('category')
df['Gender'] = df['Gender'].astype('category')
df['Embarked'] = df['Embarked'].astype('category')
```

```python
# Confirm data type conversion
df.dtypes
```

```python
# Generate summary statistics for numerical columns
df.describe()
```

```python
# Display first 5 rows of cleaned dataset
df.head()
```

```python
# Display final dataset structure and data types
df.info()
```

```python
df.to_csv('Titanic_Cleaned.csv', index=False)
```
